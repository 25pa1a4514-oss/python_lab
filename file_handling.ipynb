```python
#program on w mode
file=open("f1.txt","w")
data="hello world"
file.write(data)
print("file created")
file.close()
```

```python
#program on "r" mode
with open("f1.txt","r") as f:
  data=f.read()
  print(data)
```

```python
#program on "a" mode
with open("f1.txt","a") as f:
  f.writelines(["\napple","\nname"])
```

```python
#covering all types
with open("f1.txt","r") as f:
  print(f.readlines())
```

```python
#program that copy content from one file to another file
with open("f1.txt","r") as f:
  data=f.read()
  print("reading done")
with open("f2.txt","w") as f:
  f.write(data)
  print("file copied")
```

```python
#program that print each line in a file in reverse order
with open("f1.txt","r") as f:
  data=f.readlines()
  for i in data:
    print(i[ : :-1])
```

```python
#write a program that sort words in a file and put them in another file
with open("f1.txt","r") as f:
  data=f.read().split()
  print(data)
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
with open("f4.txt","w") as f2:
   for i in sort_data:
     f2.write(i+"\n")
print("done")
```
