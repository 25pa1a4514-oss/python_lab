```python
#write a python program to find no of duplicates in a list
a=list(map(int,input().split()))
s=set(a)
print(s)
print(len(a)-len(s))
```

```python
#write a python program to find how many numbers having duplicates in a list
L1=list(map(int,input().split()))
L2=[]
for i in L1:
  if L1.count(i)>1:
    L2.append(i)
print(L2)
s=set(L2)
print (s)
print(len(s))


```

```python
#write a python program to find non repeated elements in a list
L1=list(map(int,input().split()))
L2=[]
for i in L1:
  if L1.count(i)==1:
    L2.append(i)
print(L2)
```

```python
#write a python program that prints missing numbers between 1 to n
n=int(input())
L1=list(map(int,input().split()))
s=set(L1)
elements=set(range(1,n+1))
print(elements-s)
```
