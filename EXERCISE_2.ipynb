```python
#program for leap year or not
year=int(input("enter the year"))
if((year%4==0 or year%100==0) and (year%400!=0)):
  print(f"{year} is leap year")
else:
  print(f"{year} is not leap year")
```

```python
#example code on match case
day=int(input("enter the day"))
match day:
  case 1:
    print("monday")
  case 2:
    print("tuesday")
  case 3:
    print("wednesday")
  case 4:
    print("thursday")
  case 5:
    print("friday")
  case 6:
     print("saturday")
  case 7:
    print("sunday")
  case _:
    print("invalid")

```
