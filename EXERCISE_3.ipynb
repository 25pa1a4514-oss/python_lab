```python
#list operations
P1=[2,33,"pooji",9.4,'Q',46,14]
P2=[3,35,"HELLO",67,5,7,33]
print(P1+P2)
#repetition
print(P1 * 2)
#indexing
print(P1[5])
print(P2[3])
#SLICING
print(P1[1:5])
print(P1[1:])
print(P1[ :5])
print(P1[1:5:1])
print(P1[1:5:])
print(P1[1: : ])
print(P1[:5:])
print(P1[: : 2])
print(P1[:])
print(P1[: : -1])
#membership operator
#1st way
print(33 in P1)
print(36 not in P2)
#2nd way
if 33 in P1:
  print(" yes")
#iteration
#1st way
for i in P1:
  print(i)
#2nd qay
for i in range(0,7):
  print(P1[i])
```

```python
b1=[4,5,6,7,8,9,9,10]
b2=[2,6,8,5,4]
#buit in functions
print(len(b1))
print(sum(b1))
print(min(b1))
print(max(b1))
print(list(reversed(b1)))
print(sorted(b1))
#built in methods
#append(adding element to list)
b1.append(66)
print(b1)
#extend(adding one list to other)
b1.extend(b2)
print(b1)
#insert(index , element that need to be added in specific index)
b1.insert(3,46)
print(b1)
#pop
b1.pop()
print(b1)
#remove(deleting element)
b1.remove(10)
print(b1)
#count
b1.count(9)
print(b1)
```
