```python
#TUPLE OPERATIONS
t1=(1,2,3,4,"pooji",'A')
t2=(5,6,7,8,"HELLO")
#concatination
print(t1+t2)
#repetition
print(t1* 2)
#indexing
print(t1[5])
print(t2[3])
#SLICING
print(t1[1:5])
print(t1[1:])
print(t1[ :5])
print(t1[1:5:1])
print(t1[1:5:])
print(t1[1: : ])
print(t1[:5:])
print(t1[: : 2])
print(t1[:])
print(t1[: : -1])
#membership operator
#1st way
print(3 in t1)
print(36 not in t2)
#2nd way
if 3 in t1:
  print(" yes")
#iteration
#1st way
for i in t1:
  print(i,end=" ")
#2nd qay
for i in range(0,5):
  print(t1[i])
```
